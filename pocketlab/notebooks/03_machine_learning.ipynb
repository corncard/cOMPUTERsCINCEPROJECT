```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import sys

sys.path.insert(0, str(Path("..").resolve()))

from pocketlab.ml.clustering import cluster_spending

csv_path = Path("../data/raw/expenses.csv")
df = pd.read_csv(csv_path)
df.head()
```

```python
print(df.shape)
print(df.dtypes)
print(df.isna().sum())
```

```python
profiles, model, scaler = cluster_spending(
    df,
    n_clusters=3,
    period="W",
)
profiles.head(10)
```

```python
summary = profiles.groupby("cluster").agg(
    average_weekly_spending=("total_amount", "mean"),
    average_transactions=("transaction_count", "mean"),
    average_transaction_value=("average_amount", "mean"),
    number_of_weeks=("total_amount", "count"),
)
summary.round(2)
```

```python
for cluster_id, group in profiles.groupby("cluster"):
    plt.scatter(
        group["total_amount"],
        group["average_amount"],
        label=f"Cluster {cluster_id}",
    )

plt.xlabel("Total weekly spending (VND)")
plt.ylabel("Average transaction value (VND)")
plt.title("PocketLab: Spending Patterns")
plt.legend()
plt.show()
```
