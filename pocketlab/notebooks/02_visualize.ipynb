# Visualize spending

Create charts from the cleaned expense data.


```python
from pathlib import Path

from pocketlab.analytics import load_expenses
from pocketlab.charts import category_chart, monthly_chart

expenses = load_expenses(Path('..') / 'data' / 'raw' / 'expenses.csv')
category_chart(expenses, save_path='category_chart.png')
monthly_chart(expenses, save_path='monthly_chart.png')
```
